## Loading the Class Corpus

```python
# Corpus loading. With CORPUS = "sample" the sample corpus is used, no checks.
# With CORPUS set to a path or URL, that file is used, or a clear error is
# raised; it needs the columns below but no Quarter column, and never falls back.
# With CORPUS = "class" the class corpus is used only if its Quarter column
# matches QUARTER; otherwise the sample corpus is used.
#
#   matches, complete            use it
#   matches, incomplete          use it, warn
#   matches, missing a genre     sample corpus, say which genre
#   matches, malformed           raise
#   anything else                sample corpus
#
# The class corpus is built from Discussion 01 submissions, so early in the term
# it does not exist yet and the sample path is the normal one.
#
# A malformed class corpus raises rather than falling back, so that a problem
# with the corpus cannot be mistaken for a result.
#
# A class corpus without every genre is not used yet: the genre analyses and
# classifiers expect all of them, and would fail or mislead on fewer.
CORPUS_REQUIRED_COLUMNS = [
    "Doc_ID", "DSI_Title", "Submission File Name", "Student Name",
    "Genre of Movie", "Review Type (pos or neg)", "Movie Title", "Text",
]
CORPUS_EXPECTED_DOCUMENTS = 200  # a finished class corpus is topped up to 200;
                                 # below this it is still being assembled
CORPUS_EXPECTED_MOVIES = 5
CORPUS_EXPECTED_GENRES = ["Action", "Comedy", "Horror", "Sci-Fi"]

_corpus_cache = {}

def _is_this_quarters(df, quarter):
    """True if the corpus declares `quarter`."""
    if "Quarter" not in df.columns:
        return False
    return sorted(df["Quarter"].dropna().astype(str).unique()) == [quarter]

def _corpus_defects(df):
    """Faults that make the corpus unusable."""
    missing = [c for c in CORPUS_REQUIRED_COLUMNS if c not in df.columns]
    if missing:
        return [f"missing column(s): {', '.join(missing)}"]

    defects = []
    dupes = df["Doc_ID"][df["Doc_ID"].duplicated()].unique()
    if len(dupes):
        defects.append(f"duplicate Doc_ID(s): {', '.join(map(str, dupes[:5]))}")
    empty = df["Text"].isna() | (df["Text"].astype(str).str.len() < 100)
    if empty.any():
        ids = df.loc[empty, "Doc_ID"].astype(str).tolist()[:5]
        defects.append(f"{empty.sum()} document(s) with empty or very short Text "
                       f"(Doc_ID {', '.join(ids)})")
    return defects

def _missing_genres(df):
    """Expected genres with no documents in the corpus."""
    present = set(df["Genre of Movie"].dropna().astype(str).str.strip())
    return [g for g in CORPUS_EXPECTED_GENRES if g not in present]

def _corpus_shortfalls(df):
    """Signs the corpus is still being assembled."""
    short = []
    if len(df) < CORPUS_EXPECTED_DOCUMENTS:
        short.append(f"only {len(df)} documents")
    if df["Movie Title"].nunique() < CORPUS_EXPECTED_MOVIES:
        short.append(f"only {df['Movie Title'].nunique()} movies")
    return short

def _load_custom_corpus(path, key):
    """CORPUS set to a file path or URL: that file, or a clear error."""
    try:
        df = get_corpus_df(path)
    except Exception as exc:
        raise ValueError(
            f"Could not load CORPUS = {path!r} ({type(exc).__name__}: {exc}). "
            'Set CORPUS in the first cell to "class", "sample", or the path or '
            "URL of a corpus CSV file."
        ) from exc
    defects = _corpus_defects(df)
    if defects:
        raise ValueError(f"The corpus {path!r} cannot be used: "
                         + "; ".join(defects) + ".")
    missing = _missing_genres(df)
    if missing:
        print(f"Note: this corpus has no {' or '.join(missing)} reviews. Genre "
              f"comparisons, and Assignment 3's genre classifiers, expect all of "
              f"{', '.join(CORPUS_EXPECTED_GENRES)}.")
    print(f"Corpus: {str(path).rsplit('/', 1)[-1]} | {len(df)} documents | "
          f"{df['Movie Title'].nunique()} movies | CUSTOM")
    _corpus_cache.update(key=key, df=df, using_sample=False)
    return df.copy(), False

def load_corpus(url=None, sample_url=None, quarter=None, corpus=None):
    """Return (dataframe, using_sample). Cached until a setting changes."""
    url = CORPUS_URL if url is None else url
    sample_url = SAMPLE_CORPUS_URL if sample_url is None else sample_url
    quarter = QUARTER if quarter is None else quarter
    corpus = CORPUS if corpus is None else corpus
    bar = "=" * 74

    if not isinstance(corpus, str) or not corpus.strip():
        raise ValueError('CORPUS must be "class", "sample", or the path or URL '
                         f"of a corpus CSV file, not {corpus!r}. Set it in the "
                         "first cell.")

    key = (corpus, url, sample_url, quarter)
    if _corpus_cache.get("key") == key:
        return _corpus_cache["df"].copy(), _corpus_cache["using_sample"]

    if corpus not in ("class", "sample"):
        return _load_custom_corpus(corpus, key)

    df, mine = None, False
    if corpus == "class":
        try:
            df = get_corpus_df(url)
            mine = _is_this_quarters(df, quarter)
        except Exception:
            df, mine = None, False   # unreadable, so not this quarter's

    missing_genres = []
    if mine:
        defects = _corpus_defects(df)
        if defects:
            raise ValueError(
                f"The {quarter} class corpus is published but unusable: "
                + "; ".join(defects)
                + ". This is a problem with the corpus, not with your notebook. "
                  "Please report it on the discussion board, quoting the "
                  "notebook version printed in the first cell."
            )
        missing_genres = _missing_genres(df)

    if mine and not missing_genres:
        shortfalls = _corpus_shortfalls(df)
        if shortfalls:
            print(bar)
            print("  !   THE CLASS CORPUS LOOKS INCOMPLETE")
            print(bar)
            print(f"  It is the {quarter} corpus, but it has "
                  + " and ".join(shortfalls) + ".")
            print("  Continuing with it. It may still be being assembled, so")
            print("  re-run this cell later and check these numbers again.")
            print(bar)
        print(f"Corpus: {url.rsplit('/', 1)[-1]} | {len(df)} documents | "
              f"{df['Movie Title'].nunique()} movies | {quarter}")
        _corpus_cache.update(key=key, df=df, using_sample=False)
        return df.copy(), False

    try:
        df = get_corpus_df(sample_url)
    except Exception as exc:
        why = ("The" if corpus == "sample"
               else f"The {quarter} class corpus is not available, and the")
        raise RuntimeError(
            f"{why} sample corpus at {sample_url} could not be loaded "
            f"({type(exc).__name__}: {exc}). Check your internet connection, "
            f"then ask on the discussion board."
        ) from exc

    if corpus == "sample":
        print('Using the sample corpus, as chosen in the first cell (CORPUS = "sample").')
        print('Set CORPUS = "class" there and re-run to use your class corpus.')
    else:
        print(bar)
        print("  !!  USING THE SAMPLE CORPUS, NOT YOUR CLASS CORPUS")
        print(bar)
        if missing_genres:
            print(f"  The {quarter} class corpus is published, but it has no "
                  f"{' or '.join(missing_genres)}")
            print("  reviews yet, and this notebook compares all genres. It is still")
            print("  being assembled from Discussion 01 submissions.")
        else:
            print(f"  The published class corpus is not the {quarter} class corpus.")
            print("  This is expected early in the term: your class corpus is built")
            print("  from Discussion 01 submissions and published after they close.")
        print()
        print("  The sample corpus is a frozen teaching set. It does not contain")
        print("  your review, or anyone's from this term. Re-run this cell once")
        print(f"  your class corpus is {'complete' if missing_genres else 'published'}, "
              "and write your report from that.")
        print(bar)
    print(f"Corpus: {sample_url.rsplit('/', 1)[-1]} | {len(df)} documents | "
          f"{df['Movie Title'].nunique()} movies | SAMPLE")
    _corpus_cache.update(key=key, df=df, using_sample=True)
    return df.copy(), True
data, USING_SAMPLE_CORPUS = load_corpus()
```

### BERT Pre-Trained Multiple Classification (Genre)

Our genre classification utilizes a BERT-based uncased model to distinguish text genres. This approach capitalizes on BERT's deep semantic understanding from uncased text, aiming for genre identification.

https://huggingface.co/google-bert/bert-base-uncased

```python
def bert_classifier_multiple(texts, labels):
    # Encode labels
    encoder = LabelEncoder()
    encoded_labels = encoder.fit_transform(labels)

    # Split the dataset
    X_train, X_test, y_train, y_test = train_test_split(texts, encoded_labels, test_size=0.10, random_state=42)

    # Load a tokenizer and model suited for your multi-class task
    #tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
    #tokenizer = AutoTokenizer.from_pretrained("bert-large-uncased")
    tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
    #model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=len(encoder.classes_))
    #model = AutoModelForSequenceClassification.from_pretrained("bert-large-uncased", num_labels=len(encoder.classes_))
    model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=len(encoder.classes_))

    # Initialize the pipeline
    classifier = pipeline("text-classification", model=model, tokenizer=tokenizer)

    # Perform classification
    results = classifier(X_test, truncation=True)

    # Decode predictions to original labels
    predictions = [int(result['label'].split('_')[-1]) for result in results]
    decoded_predictions = encoder.inverse_transform(predictions)

    # Calculate accuracy
    decoded_y_test = encoder.inverse_transform(y_test)
    accuracy = accuracy_score(decoded_y_test, decoded_predictions)
    print(f"Multi-class Classification Accuracy: {accuracy}")
    return accuracy

```

### Confusion Matrix

This code snippet generates a confusion matrix to visually evaluate the performance of our classification model by comparing the actual versus predicted labels.

```python
from sklearn.metrics import confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline

def bert_classifier_multiple(texts, labels):
    # Initialize and fit the label encoder
    encoder = LabelEncoder()
    encoded_labels = encoder.fit_transform(labels)

    # Split the dataset
    X_train, X_test, y_train, y_test = train_test_split(texts, encoded_labels, test_size=0.10, random_state=42)

    # Load a tokenizer and model suited for the multi-class task
    tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
    #tokenizer = AutoTokenizer.from_pretrained("bert-large-uncased")

    model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=len(encoder.classes_))
    #model = AutoModelForSequenceClassification.from_pretrained("bert-large-uncased", num_labels=len(encoder.classes_))

    # Initialize the pipeline
    classifier = pipeline("text-classification", model=model, tokenizer=tokenizer)

    # Perform classification
    results = classifier(X_test, truncation=True)

    # Process predictions
    predictions = [int(result['label'].split('_')[-1]) for result in results]

    # Decode predictions and true labels back to original labels
    decoded_predictions = encoder.inverse_transform(predictions)
    decoded_y_test = encoder.inverse_transform(y_test)

    # Return the true labels, predicted labels, and the encoder (for plotting confusion matrix)
    return decoded_y_test, decoded_predictions, encoder

# Assuming `data["Genre of Movie"]` is your labels array
true_labels, predicted_labels, encoder = bert_classifier_multiple(final_processed_text, data["Genre of Movie"])
```

```python
def plot_confusion_matrix(true_labels, predicted_labels, classes):
    cm = confusion_matrix(true_labels, predicted_labels, labels=classes)
    plt.figure(figsize=(12, 12))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
    plt.title('Confusion Matrix')
    plt.ylabel('True Labels')
    plt.xlabel('Predicted Labels')
    plt.show()

plot_confusion_matrix(true_labels, predicted_labels, encoder.classes_)
```
